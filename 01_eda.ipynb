Step 2: Create a Virtual Environment (.venv)
Using a virtual environment keeps your project dependencies (Pandas, SQLAlchemy, BigQuery drivers) isolated so they don't mess up your system Python packages.


    1.	Open the built-in terminal in VS Code: Terminal  New Terminal (or press Ctrl + ` / Cmd + `).

    2.	Run the command to create a virtual environment:

        Run standard python3:
        Bash
        python3 -m venv .venv

        Step-by-Step Activation for Linux / WSL
            1.	Create the environment:
            Bash
            python3 -m venv .venv
            (Note: If Linux tells you No module named venv, install it using sudo apt update && sudo apt install python3-venv).
            2.	Activate the environment:
            Bash
            source .venv/bin/activate
            (Notice that on Linux/WSL, it uses source .venv/bin/activate instead of .venv\Scripts\Activate.ps1).
            3.	Install your packages:
            Bash
            pip install pandas sqlalchemy google-cloud-bigquery db-dtypes matplotlib seaborn jupyter
    You will see (.venv) appear at the front of your terminal prompt, confirming your virtual environment is active!

```python
from sqlalchemy import create_engine
import pandas as pd

# 1. Configuration
GCP_PROJECT_ID = "ds6-module-2-olist-analytics"  # Replace with the project ID Gina shared
DATASET_NAME = "ingestion"        # Replace with Jamie's target dataset; temporarily using Gina's dataset for testing

# 2. Build SQLAlchemy Engine
engine = create_engine(f"bigquery://{GCP_PROJECT_ID}/{DATASET_NAME}")

# 3. Test Query using Gina's table names (products, orders, order_items)
test_query = """
    SELECT 
        p.product_id,
        p.product_category_name,
        i.price
    FROM products p
    JOIN order_items i ON p.product_id = i.product_id
    LIMIT 10
"""

# Load directly into Pandas
df_mart = pd.read_sql(test_query, con=engine)
print("Successfully connected! Preview of category_performance_monthly:")
print(df_mart.head())
```
